```python
import pandas as pd
import re

df = pd.read_csv('metadata.csv')  # Or pd.read_json('metadata.json')
df = df.rename(columns={'param_name': 'parameter', 'default_value': 'input_value', 'input_type': 'param_type'})
df.head()
```

```python
# The apply lines here...
df['input_length'] = df['input_value'].apply(lambda x: len(str(x)))
df['num_special_chars'] = df['input_value'].apply(lambda x: len(re.findall(r'[^a-zA-Z0-9\s]', str(x))))
df['num_digits'] = df['input_value'].apply(lambda x: len(re.findall(r'\d', str(x))))
df['contains_sql_keyword'] = df['input_value'].apply(lambda x: bool(re.search(r'(select|union|drop|insert|delete|update)', str(x), re.IGNORECASE)))
df['contains_html_tag'] = df['input_value'].apply(lambda x: bool(re.search(r'<[^>]+>', str(x))))

columns = ['url', 'parameter', 'input_value', 'input_length', 'num_special_chars',
           'num_digits', 'param_type', 'contains_sql_keyword', 'contains_html_tag']
features_df = df[columns]
features_df.to_csv('features_extracted.csv', index=False)
features_df.head()
```
