```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load features
df = pd.read_csv('features_extracted.csv')
```

```python
# 1. Histogram of input lengths
plt.figure(figsize=(10, 6))
sns.histplot(df['input_length'], bins=10, kde=True)
plt.title('Histogram of Input Lengths')
plt.xlabel('Input Length')
plt.ylabel('Frequency')
plt.show()  # Expect peaks at 0, 5 (Login), 32 (tokens)
```

```python
# 2. Bar chart of parameter types
plt.figure(figsize=(10, 6))
sns.countplot(x='param_type', data=df)
plt.title('Bar Chart of Parameter Types')
plt.xlabel('Parameter Type')
plt.ylabel('Count')
plt.show()  # Shows counts for text, password, submit, hidden
```

```python
# 3. Correlation heatmap (numeric features)
numeric_cols = ['input_length', 'num_special_chars', 'num_digits',
                'contains_sql_keyword', 'contains_html_tag']
corr = df[numeric_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Numeric Features')
plt.show()  # In your data, correlations might be low due to simple values
```
